## La función `print()`

Esta función permite mostrar un texto especifico en la pantalla. El texto a mostrar (cadena de texto) se escribe como argumento de la función.


image.png

---
[Tip]: Para ejecutar el código de Python en la celda de código a continuación, haga clic en la celda para seleccionarla y presionar `CTRL` + `Enter`.

---


```python
# Probemos el primer output en Python

print("¡Hola Mundo!")
```

---
[Tip:] `print()` es una función. En este caso, recibe el string (cadena de texto) `'¡Hola Mundo!'` como argumento, para indicarle a Python qué debe imprimir como output.

---


```python
# Un argumento puede escribirse con comillas dobles o simples

print("Pensamiento Computacional")
print('Pensamiento Computacional')
```

La función `print()` admite varios argumentos seguidos. En el programa, los argumentos deben separarse por comas.

```python
# La coma que separa cada argumento representa un espacio vacío

print("Hola", "Adiós") # es igual a print("Hola Adiós")
```

```python
# Sin una coma que separe argumentos se interpreta como un solo argumento sin espacio entre ellos

print("Hola" "Adiós") # es igual a print("HolaAdiós")
```

Al final de cada `print()`, Python añade automáticamente un salto de línea.

```python
# Imprimamos dos funciones con print() en dos líneas de código
print("Aula")
print("Z105")
```

```python
# Agregamos un print() sin un argumento, ¿qué sucede?
print("Aula")
print("")
print("Z105")
```

Si se quieren separar los argumentos en la salida, hay que incluir los espacios deseados.

```python
# Se puede agregar un espacio vacío al final del argumento de la primera línea
print("Aula ", end="")
print("Z105") # el resultado es igual a print("Aula", "Z105") o print("Aula Z105")
```

```python
print("Aula", end="")
print("Z105")
```

También se puede ejecutar dos argumentos y especificar un separador.

```python
# Se puede indicar qué tipo de separador deseas entre los argumentos, estos se imprimiran en vez de indicar un espacio vacío por defecto
print("Mi nombre es", "Luisa", sep="--")
print("Mi nombre es", "Luisa", sep=":")
print("Mi nombre es", "Luisa", sep=": ")
```

## ¿Cómo definimos una variable?


Una variable es un nombre que hace referencia a un valor almacenado en la memoria del programa. Asimismo, las variables permiten reutilizar información sin necesidad de escribirla varias veces.

```python
frase = "¡Bienvenidos al semestre 2026-I!"
curso = "Fotografía"
créditos = 3
nota = 11.5

print(curso)
print(créditos)
print(nota)
print(frase)
```

El contenido de una variable no es permanente. Podemos modificarlo en cualquier momento.

Python siempre conserva el valor más reciente asignado a la variable.

```python
print(curso)

curso = "Pensamiento Computacional"

print(curso)
```

```python
mensaje = "Me gustan los perros"
print(mensaje)
mensaje = "Me gustan los gatos"
print(mensaje)
```

```python
print(mensaje)
```

```python
cantante = "Taylor Swift"
print("cantante")
print(cantante)
```

## Restricciones en los nombres de variables
Al crear variables es importante seguir algunas reglas.

1. Solo pueden contener
* letras
* números
* guion bajo (_)
2. No pueden comenzar con un número
3. No pueden ser palabras reservadas (nombres de funciones o métodos de Python)
4. Utilicen nombres descriptivos

```python
#¿Qué sucede aquí?
CR_7 = 965
print(CR_7 + 1)
```

```python
# ¿Qué sucede aquí?
CR = "965"
print(CR + 1)
```
