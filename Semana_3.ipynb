# **1. Expresiones booleanas**
Los valores booleanos son **`True`** o **`False`**. Un valor booleano también puede ser el resultado de evaluar una expresión.

````python
print(type(False))
print(type(True))
````

````python
resultado = 10 > 5

print(resultado)
print(type(resultado))
````

En Python, una expresión booleana es una expresión que, al ser evaluada durante la ejecución del código, produce uno de dos valores posibles: **`True`** o **`False`**.

Estos valores pertenecen al tipo de dato bool y se utilizan para representar si una condición es verdadera o falsa.

El resultado de una expresión booleana puede asignarse a una variable:

````python
# Convertir False en int
print(int(True))
print(int(False))
````

````python
# Convertir True en float
print(float(False))
print(float(True))
````

# **2. Operadores booleanos**

### **Operadores comparativos**


Operador | Significado |True	| False
--- | --- | --- | ---
**==**	| igual a	| 2 **==** 2 | 2 **==** 3
**!=**	| distinto a | 3 **!=** 2 | 2 **!=** 2
**<**	| menor que	| 2 **<** 13 | 2 **<** 2
**<=**	| menor o igual que |	2 **<=** 2 | 3 **<=** 2
**>**	| mayor que	| 13 **>** 2 | 2 **>** 13
**>=**	| mayor o igual que	| 3 **>=** 2	| 2 **>=** 3

¡Recuerda que el signo **`=`** se reserva para asignación! Las expresiones booleanas observan variables pero nunca las cambian.

````python
# ¿Qué sucede aquí?

print("casa" < "ca") # diferencia en la cantidad caracteres
print("b" < "aa") # diferencia está en valor asignado por el unicode del caracter
print("c" < "cc")
print("za" < "az")
````

Si desean revisar más sobre los unicode (https://docs.python.org/es/3/howto/unicode.html#introduction-to-unicode)

````python
print(2 < 5)
print(2 <= 5)
print(3 > 7)
print(3 >= 7)
print(3 == 3)
print("Publicidad" == "Publicidad")
print("Comunicación Audiovisual" != "Comunicación para el Desarrollo")
````

````python
profesor_1 = "Joseph Crawford"
profesor_2 = "Eduardo Villanueva"

print(profesor_1 != profesor_2)
````

````python
mascota = "perro"
print(mascota)

mascota = "gato"

print(mascota == "perro")
````

````python
día_ayer = "jueves"
día_hoy = "viernes"

días_iguales = int(día_ayer == día_hoy)


evaluar_igualdad = (1 == días_iguales)

print(días_iguales)
print(evaluar_igualdad)
print("El resultado de la comparación de los valores de las variables es", evaluar_igualdad)
````

### **Operadores de pertenencia (in/ not in)**


Operador | Función | True | False
--- | --- | --- | ---
**in**	| el objeto de la izquierda es miembro del objeto de la derecha	| "c" **in** "comunicación"	| "e" **in** "comunicación"
**not in** | el objeto de la izquierda NO es miembro del objeto de la derecha	| "e" **not in** "comunicación"	| "c" **not in** "comunicación"

````python
## Creamos una variable cuyo valor sea un string
mensaje = 'Los documentos se recojen los días laborables'

# Revisa si "documentos" está en la variable mensaje
print("documentos" in mensaje)

# Revisa si "hello" está en la variable mensaje
print('hello' not in mensaje)
````

El operador **`in`** se utiliza para verificar si un valor está presente en una secuencia, como una lista o cadena de texto. Sin embargo, en el siguiente caso, el dato del tipo entero o float no son iterables.

````python
print(1 in 10)
````

````python
print('1' in '10')
````

## **Operadores lógicos (and, or y not)**


Operación	| Función	| True | False
--- | --- | --- | ---
bool1 **and** bool2	| Es True si tanto bool1 como bool2 son True, de lo contrario False |	(5 == 5 **and** 3 < 5) |	(5 == 5 **and** 3 > 5)
bool1 **or** bool2 | Es True cuando al menos una de las expresiones booleanas es True, de lo contrario False	| (5 == 5 **or** 3 > 5) | (5 != 5 **or** 3 > 5)
**not** bool1 | Es True si bool1 es False, de lo contrario False	| (**not** 5 != 5) | (**not** 5 == 5)

````python
print(3>5)
````

````python
print(not 3>5) # not False
````

````python
curso = "fotografía"
print(not curso.endswith("a"))
print(not curso.startswith("o"))
````

# **3. Función `input()`**
Muchas aplicaciones necesitan que el usuario ingrese datos.

Para ello utilizamos la función `input()`, que permite leer información escrita desde el teclado.

```
variable = input("Mensaje para el usuario: ")
```

Cuando el programa ejecuta a input():

* Muestra un mensaje en la pantalla.
* Espera a que el usuario escriba una respuesta.
* Guarda esa respuesta en una variable.

**Importante:** La función input() siempre devuelve un texto (str), aunque el usuario escriba números.

````python
codigo = input("Escribe tu nombre: ")
print("Bienvenido,", codigo)
````

````python
correo = input("Por favor ingresa tu correo: ")
print(correo)
````

````python
print(type(codigo))
print(type(correo))
````

````python
print("¿De qué carrera eres?")
carrera = input()
print("Estos son los cursos de este semestre en", carrera)
````

````python
carrera = input("¿De qué carrera eres? ")
print("Estos son los cursos de este semestre en", carrera)
````
