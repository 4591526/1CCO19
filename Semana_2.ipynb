# **1. Formato de cadenas con `f-strings`**

Los **f-strings** son una forma de escribir cadenas de texto (strings) en Python que permiten insertar variables o expresiones directamente dentro de la cadena usando llaves `{}`. Y al ejecutar tu código, todos los nombres de las variables serán remplazados con sus respectivos valores.

Esto facilita la lectura del código y hace más sencilla la construcción de mensajes dinámicos.

Para crear f-strings, solo tienes que agregar la letra f antes de las comillas.

```python
print("Estoy llevando la siguiente cantidad de cursos:", numero_cursos, ".")
print(f'Estoy llevando la siguiente cantidad de cursos:{numero_cursos}.')
```

##**Integers**
Los números enteros pueden ser números negativos o positivos.

```python
# Suma de dos integers
numero_1 = 7
numero_2 = 9

print(numero_1 + numero_2)
print(7 + 9)
```

```python
# Resta de dos integers
print(numero_1 - numero_2)
print(type(numero_1 - numero_2))
```

```python
# Comprueba el tipo de dato de la multiplicación de los resultados anteriores
print(numero_1 * numero_2)
print(type(numero_1 * numero_2))
```

## **Floats**
Floats representan números reales; son un superconjunto de números enteros pero también incluyen "números decimales".



```python
nota_PC1 = 10.5
nota_PC2 = 16.5

promedio = (nota_PC1 + nota_PC2)/2
print(promedio)
print(type(promedio))
```

# **3. Conversión de datos**
**Convertir integers en floats:** Cuando convertimos un integer en un float, en realidad no cambiamos el valor del número.

```python
# Convirtamos 2 en un float

print(float(2))
```

**Convertir floats en integers**: Si convertimos un float en un integer, potencialmente podríamos perder algo de información.

```python
# Convirtamos 1.1 en integer

print(int(1.1))
```

**Convertir de strings a integers o floats**: A veces, podemos tener un string que contiene un número dentro de ella. Si este es el caso, podemos convertir ese string que representa un número en un entero usando `int()`:

```python
# De string a integer

print(int('1'))
```

```python
numero = "20"
numero_convertido = int(numero)
print(type(numero))
print(numero)
print(type(numero_convertido))
print(numero_convertido)
```

```python
# Convirtamos un string en un integer con error

int('1 or 2 people')
```

Ten en cuenta que las cadenas se pueden representar con comillas simples ('1.2') o doble comillas ("1.2"), pero nunca combinados ("1.2').

## **Operador de asignación `+=`**

El operador `+=` se utiliza para actualizar el valor de una variable sumándole otro valor.

```python
seguidores = 100

seguidores = 200

print(seguidores)
```

```python
seguidores = 100

seguidores += 100
seguidores += 500
seguidores += 1000

print(seguidores)
```

Cada vez que utilizamos `+=`, la variable conserva su valor actualizado.

```python
# ¿Qué sucede aquí?
# Inicio con una cantidad de seguidores
seguidores = 100

# Más adelante, mi número de seguidores va creciendo
seguidores = seguidores + 100

# Pasan un tiempo y sigue creciendo el número de seguidores
seguidores = seguidores + 500

# Ya llevo 1 año y el número de seguidores sigue creciendo
seguidores = seguidores + 1000

# ¿Cuántos tengo?
print(seguidores)
```
