# **1. Slicing**

Slicing, o estructura de rebanado, permite seleccionar una parte de una secuencia, como una cadena (str) o una lista (list), mediante sus índices.

```
# La estructura es:
secuencia[inicio:fin:paso]
```
* inicio = dónde comienzo
* fin = dónde paro (sin incluirlo)
* paso = cuánto avanzo o retrocedo


````python
# ¿Qué sucede aquí?

mascota = "perro"

print(mascota[0])
````

````python
print(mascota[0:1])
````

````python
print(mascota[0:2])
````

````python
print(len(mascota))
print(mascota[0:4])
````

````python
print(mascota[0:5:1])
print(mascota[0:5:2])
````

````python
print(mascota[-5:-1])
print(mascota[-5:])
````

````python
print(mascota[-1:-6:-1])
````

````python
print(mascota[::-1])
````

## **4. Declaraciones condicionales en Python**
Las declaraciones condicionales permiten que un programa tome decisiones según se cumpla o no una determinada condición.

La condición se evalúa como una expresión booleana, cuyo resultado puede ser **True** o **False**. Dependiendo de este resultado, Python ejecuta un bloque de código o continúa con la siguiente instrucción.

### **Instrucciones if**
La **declaración if** permite ejecutar un bloque de código únicamente cuando se cumple una condición.

image.png

````python
cantidad = 5 # Prueba cambiando el valor de cantidad
if cantidad <= 5:
    print(cantidad)
````

````python
numero_cualquiera = 10.50 # Prueba cambiando el valor de numero_cualquiera
if numero_cualquiera == 10:
    print("El número es igual a 10.")
if numero_cualquiera > 9:
    print("El número es mayor que 9.")
if numero_cualquiera >= 10:
    print("El número es mayor o igual a 10.")
if numero_cualquiera != 11:
    print("El número no es igual a 11.")
````

---
Recuerda que si tenemos estructura **if -if**, Python revisará cada una de las declaraciones y evaluará cuales son **True**.

---

### **Decisiones de dos escenarios: if-else**

La estructura **if-else** permite que un programa elija entre dos caminos posibles según el resultado de una condición.

* Si la condición es True, se ejecuta el bloque de código del if.
* Si la condición es False, se ejecuta el bloque de código del else.

image.png

````python
palabras_articulo = 850  # Prueba cambiando el valor de palabras_articulo

if palabras_articulo >= 1500:
    print("El artículo está listo para ser enviado.")
else:
    print("El artículo es muy corto, por favor amplía el contenido.")
````

### **Decisiones con múltiples opciones: if-elif-else**
La estructura **if-elif-else** permite evaluar varias condiciones de manera secuencial. Python comprueba las condiciones en el orden en que aparecen y ejecuta únicamente el primer bloque cuya condición sea verdadera.

* Si ninguna de las condiciones del if o de los elif se cumple, se ejecuta el bloque else.

* Una estructura puede contener un if, uno o varios elif y, opcionalmente, un else.

image.png

````python
numero_3 = 0  # Prueba cambiando el valor de numero_3

if numero_3 > 0:
    print('El número es positivo.')

elif numero_3 < 0:
    print('El número es negativo.')

else:
    print('Cero')

print('Este mensaje siempre se ejecutará.')
````

---
Recuerda que si tenemos declaraciones **if-elif**, si una condición resulta en True, no verificará las otras condiciones.

## **Sintaxis de las condicinales**
### **Indentación**
Es el espacio al inicio de cada línea de código para indicar qué líneas pertenecen a un bloque.

* En Python se usa generalmente 4 espacios o un Tab.
* La indentación es obligatoria y define la estructura del código.

````python
print("¡Bienvenidos al clásico!") # Esta línea de código está fuera de las declaraciones condicionales

goles_real_madrid = 2
goles_barcelona = 1

if goles_real_madrid > goles_barcelona:
    print("¡Hala Madrid!")
elif goles_barcelona > goles_real_madrid:
    print("¡Més que un club!")
else:
    print("Ninguno de los equipos se llevó la victoria.")

print("¡Nos vemos en el próximo clásico!") # Esta línea de código está fuera de las declaraciones condicionales
````

### **Anidación**

Consiste en colocar una estrutura condicional dentro de otra.

````python
x = float(input("Ingresa un número: "))
if x >= 0:
    if x == 0:
        print("Cero")
    else:
        print("Número positivo")
else:
    print("Número negativo")
````

````python
# ¿Qué pasa aquí?

x = float(input("Ingresa un número: "))
if x >= 0:
    elif x == 0:
        print("Cero")
    else:
        print("Número positivo")
else:
    print("Número negativo")
````
